```python
import numpy as np
import seaborn as sb
import matplotlib.pyplot as plt
import scipy as sp
import wave
import IPython
```

## Read the wave file first thing

```python
def get_audio_data(name):
    audioData = None
    audioParams = None
    with open(name, 'rb') as f:
        wover = wave.Wave_read(f)
        frames = wover.getnframes()
        audioData = wover.readframes(frames)
        audioParams = wover.getparams()
    return audioData, audioParams, IPython.display.Audio(name)
```

```python
data, params, display = get_audio_data('penopticon.wav')
```

## Divide the timeblock into 1s chunks

```python
samples = np.frombuffer(data, dtype=f'int{8*params.sampwidth}') / 2**(8 * params.sampwidth)
samples
```

```python
chunk_count = np.ceil(params.nframes/params.framerate)*params.framerate - params.nframes
chunks = np.pad(samples, (0, int(chunk_count))).reshape((-1, params.framerate))

def get_block(n):
    if n % 2 == 0:
        return chunks[n//2]
    else:
        i = n // 2
        blk = params.framerate
        return np.concat([chunks[i][blk//2:blk], chunks[i+1][:blk//2]]) 
```

```python
# Since we had already implemented this we are just gonna use the library function this time

window_size = 31 * params.framerate // 1000    # 31 ms window
hop_size = window_size // 2
sampling_frequency = 1 / params.framerate

stft = sp.signal.ShortTimeFFT(sp.signal.get_window('hann', window_size), hop_size, sampling_frequency)
```

## Defining the equations mentioned in the research paper using code just incase
- get the amplitudes of the sound signal where
- get the STFT of the sound signal
- AV is the expectation of the signal
- RSD is the standard deviation

```python
# just getting somewhat of a interpolated value for the audio signal, linear interpolation
def s(i, t):
    blk = get_block(i)
    j = t * len(blk)
    k = int(j)

    if j - k == 0 and j + 1 <= len(blk):
        return blk[k]
    else:
        interp = j - k
        return blk[k] * (1 - interp) + blk[k+1] * interp
```

```python
def AV(blk):
    return np.sum(blk) / params.framerate
```

```python
def RSD(blk):
    mean = AV(blk)
    v = blk - mean
    return np.sqrt(np.sum(v * v) / params.framerate) / mean
```

## Calculate all the 12 parameters
- Amplitude
- Spectral Width
- Low-Freuency Power
- High-Frequency Power
- Fundamental Frequency
- Harmonicity

```python
# Amplitude
def Pa(i):
    blk = get_block(i)
    return AV(np.abs(blk))

def Ps(i):
    blk = get_block(i)
    return RSD(np.abs(blk))
```
